Dictionaries

```python
m = {"a": 1, "b": 2, "c": 3}

print(m)
print(type(m))
print(m["a"])
m["a"] = 10
print(m)
```

Dictionaries methods

```python
z = {"a": 1, "b": 2, "c": 3}
print(z.keys())
print(z.values())
print(z.items())
print(z.get("a"))
print(z.pop("a"))
print(z)
```

Dictionaries Comprihensions

```python
table = [5*i for i in range(1, 11)]

print("Multiplication Table of 5:", table)
```
